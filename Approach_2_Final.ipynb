```python
import pandas as pd
import numpy as np
from tqdm import tqdm
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("/content/drive/MyDrive/cicids2017_curated_38cols_Binary_Latest.csv")
df
```

```python
X = df.drop(columns=['Label'])
Y = df['Label'].values
```

```python
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.2,random_state=0)
```

```python
from tensorflow.keras.utils import to_categorical

Y_train = to_categorical(Y_train, num_classes=2)
Y_test = to_categorical(Y_test, num_classes=2)
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

X_train_trf = scaler.fit_transform(X_train)
X_test_trf = scaler.transform(X_test)
```

```python
import tensorflow
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense
```

```python
# model 1
model = Sequential()
model.add(Dense(32,activation = 'relu',input_dim = 37))
model.add(Dense(32, activation='relu'))
model.add(Dense(16, activation='relu'))
model.add(Dense(16, activation='relu'))
model.add(Dense(8, activation='relu'))
model.add(Dense(2, activation='softmax'))
model.summary()

model.compile(loss='categorical_crossentropy',optimizer = 'Adam',metrics=['accuracy'])
```

```python
# model 2
model = Sequential()
model.add(Dense(32,activation = 'relu',input_dim = 37))
model.add(Dense(32, activation='relu'))
model.add(Dense(16, activation='relu'))
model.add(Dense(16, activation='relu'))
model.add(Dense(8, activation='relu'))
model.add(Dense(2, activation='softmax'))
model.summary()

model.compile(loss='categorical_crossentropy',optimizer = 'SGD',metrics=['accuracy'])
```

```python
# model 3
model = Sequential()
model.add(Dense(32,activation = 'relu',input_dim = 37))
# model.add(Dense(32, activation='relu'))
# model.add(Dense(16, activation='relu'))
# model.add(Dense(16, activation='relu'))
model.add(Dense(8, activation='relu'))
model.add(Dense(2, activation='softmax'))
model.summary()

model.compile(loss='categorical_crossentropy',optimizer = 'Adam',metrics=['accuracy'])
```

```python
model.fit(X_train_trf,Y_train,epochs = 100,batch_size = 1024,validation_split = 0.1)
```

```python
Y_true = np.argmax(Y_test, axis=1)
Y_pred = np.argmax(model.predict(X_test_trf), axis=1)
```

```python
from sklearn.metrics import classification_report

print(classification_report(Y_true, Y_pred))
```
